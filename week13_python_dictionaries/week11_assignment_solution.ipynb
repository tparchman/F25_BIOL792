## A quick bioinformatics exercise using dictionaries

For this we are going to use a genome annotation file for *S. cerevisiae* `Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3`

General goals:
Use a real GFF3 genome annotation file to:
- Parse the file line by line in Python
- Use a dictionary where:
- keys = feature types (column 3, e.g. gene, CDS, tRNA, …)
- values = counts of how many times each feature appears

Using dictionaries can be useful to answer questions like:

- “How many genes are annotated?”
- “How many CDS records are there?”
- “What are all the feature types present in this annotation?”

Here we are going to keep track of the unique features (using keys of a dictionary) and then count how many times each type of feature is annotated to a region of the genome (using the values of that dictionary, think `dict[key] += 1`)

```python
feature_counts = {}

with open("Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3", "r") as IN:
    for line in IN:
        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")
        if len(fields) < 3:
            continue

        feat_type = fields[2]

        # Use .get() to provide a default of 0 if key doesn't exist yet
        feature_counts[feat_type] = feature_counts.get(feat_type, 0) + 1
```

```python
for i, k in feature_counts.items():
    print("feature type,", i, "number of appearances:", k)
```
