## Dictionaries

Dictionaries (also known as associative arrays, or hashes) are data structures that store unordered pairs of objects. Each pair consists of a key and a value. These types of structures that store pairs of data are common to many programming languages for several reasons. First, we often want one piece of information to be directly connected to another (names and phone numbers; gene name and sequence; site identifier and measurement; codons and their associated amino acids). We can pair such data using lists, but this turns out to be quite slow (and clumsy) for larger data because lists have a fixed ordering of elements. Dictionaries are unordered, thus they have the advantage of allowing you to look up or access pairs of information much more quickly.

Dictionaries are great data structures because they let you organize data using meaningful labels (keys) instead of positions. Additionally, they make your code cleaner, faster, and much easier to understand.

A **dictionary** is a mapping between **keys** and **values**:

- **Keys** — unique identifiers (names, IDs, sequence labels)  
- **Values** — the information attached to each key (numbers, strings, lists, other dicts, etc.)


### Why dictionaries matter

Dictionaries are:

- **Fast** — constant-time lookups (`mydict[key]` is instant)
- **Flexible** — values can be any Python object
- **Clear** — keys give meaning to data (unlike list indexes)
- **Everywhere** — used internally by Python, pandas, JSON, APIs, and more

In bioinformatics and data science, dictionaries make code:

- **More readable** → `gene_counts["COX1"]` instead of `gene_counts[42]`
- **More robust** → order-independent, so your code doesn’t break
- **More expressive** → they naturally model key→value relationships (e.g., accession → sequence)

## Useful dictionary methods

Before we think about building dictionaries while processing real data, let’s use the example dictionary from above to demonstrate some commonly used and highly practical methods for extracting, updating, and manipulating information. These will come up constantly when you start working with real datasets.

Below are the methods we will focus on:

- **`.pop(key)`** – Removes a key–value pair and *returns* the value.  
  Useful when you want to extract something and delete it from the dictionary at the same time.

- **`.get(key, default=None)`** – Returns the value for `key`, but *does not raise an error* if the key is missing.  
  You can optionally supply a default value. This is safer than `mydict[key]`.

- **`.items()`** – Returns key–value pairs as tuples.  
  Essential when looping through both keys and values.

- **`.keys()`** – Returns all keys in the dictionary.  
  Helpful for checking membership or iterating through key names.

- **`.values()`** – Returns all keys in the dictionary.  
  Helpful for checking membership or iterating through values.

- **`.update(other_dict)`** – Merges another dictionary into the current one.  
  If keys overlap, the new values overwrite the old ones.

- **`.copy()`** – Makes a shallow copy of the dictionary so you can modify it without changing the original.

## Nested Dictionaries

A nested dictionary is simply a dictionary that contains one or more dictionaries as its values.  This structure is extremely useful when your data naturally has multiple layers — for example:

- gene → sample → expression value  
- population → individual → metadata  
- species → chromosome → sequence length  
- team → position group → players

Nested dictionaries add flexibility and let you represent hierarchical data cleanly, but they can also become more complex to navigate. The examples below illustrate how they work.

### Creating a nested dictionary


````python
Bengals = {
    'Offense': {
        9:  'Burrow',
        1:  'Chase',
        86: 'Fant',
        75: 'Brown',
        5:  'Higgins'
    },
    'Defense': {
        29: 'Taylor-Britt',
        55: 'Wilson',
        97: 'Stewart',
        21: 'Hilton',
        20: 'Stone'
    }
}
````

The top-level keys are "Offense" and "Defense" and each of those maps to another dictionary:
- Keys = jersey numbers
- Values = player names

````python
# Accessing element using key 
print("Offense dict: ", Bengals['Offense']) 
print("Offense, number 1: ", Bengals['Offense'][1]) 
print("Defense, number 29: ", Bengals['Defense'][29]) 

#iterating through nested dictionaries
for i in Bengals:
    print(i, Bengals[i])
    print(i, Bengals[i].keys())
    print(i, Bengals[i].values())
````

Lets look at some example code below that reads in a .fasta file containing alternating lines of identifiers and DNA sequence data. Within the `for` loop, we process two lines at a time, assigning them to different variables. We then build the dictionary "ID_Seq" within the loop, using ID as keys and Seq as values. I like to name dictionaries with two part names that suggest the variables used for keys and values, but like scalars and lists, you can name dictionaries anything you want.

````python
ID_Seq={}           # initializing the ID_seq dictionary
with open("SG_ref.fasta.short") as IN:
    for Line in IN:
        ID = Line.strip('\n')
        Seq = IN.readline()
        Seq = Seq.strip('\n')
        ID_Seq[ID]=Seq      # building the ID_seq dictionary
#WORK ON DICTIONARY OUTSIDE OF LIST
````

### Here is another example of building a dictionary, but this time based on a single string counting bases in a DNA sequence

The code below loops through `seq`, which is a string of DNA bases, assigns the `counts` dictionary key as the character (`base`), and updates the value for each base as it encounters them. So, the first time `A` is encountered, A is key and 1 is set to value. Second time A is encountered, value is set to `+=1`, and so on.

````python
seq = "ATGCTTACGAATTCGACCCGTAATATATATATAT"

counts = {}   # empty dictionary

for base in seq:
    if base not in counts:
        counts[base] = 1
    else:
        counts[base] += 1

print(counts)
````

## Using Dictionaries in Bioinformatic workflows

Dictionaries are extremely powerful in bioinformatics because so much biological data is naturally “keyed”:
- Gene → metadata
- Chromosome → list of features
- Transcript → exons
- Sequence ID → sequence
- Feature type → count
- Sample name → summary statistics

Below are several realistic examples using a GFF3 annotation file for Saccharomyces cerevisiae chromosome II.

You can download it here (or just use from the course github directory):
https://ftp.ensembl.org/pub/release-92/gff3/saccharomyces_cerevisiae/Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3.gz

### Example 1: Mapping gene IDs to their coordinates

````python
gene_coords = {}

with open("Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3") as fh:
    for line in fh:
        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")
        ftype = fields[2]

        if ftype == "gene":
            chrom = fields[0]
            start = int(fields[3])
            end   = int(fields[4])
            attr = fields[8]

            gene_id = [x.split("=")[1] for x in attr.split(";") if x.startswith("ID=")][0]
            gene_coords[gene_id] = (chrom, start, end)
    for i, k in gene_coords.items():
        print(i,k)
````

### Example 2: Creating a dictionary of gene (key) to list of CDS intervals (value)

Useful for modeling exon/CDS structure.

````python
cds_dict = {}

with open("Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3") as fh:
    for line in fh:
        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")
        ftype = fields[2]

        if ftype == "CDS":
            start = int(fields[3])
            end   = int(fields[4])
            attr = fields[8]

            parent = [x.split("=")[1] for x in attr.split(";") if x.startswith("Parent=")][0]

            if parent not in cds_dict:
                cds_dict[parent] = []

            cds_dict[parent].append((start, end))
    for i, k in cds_dict.items():
        print(i, k)
````

### Example 3: Feature type (key) list of lengths (values)

Great for plotting distributions later.

````python
lengths = {}

with open("Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3") as fh:
    for line in fh:
        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")
        ftype = fields[2]
        start, end = int(fields[3]), int(fields[4])
        length = end - start + 1

        lengths.setdefault(ftype, []).append(length)
        
    print(len(lengths["gene"]), "gene lengths collected")
    for i, k in lengths.items():
        print(i, k)
````

### Example 4: Dictionary-driven filtering (real-world task)

Code below will return all genes longer than 2000 bp

````python
gene_coords = {}

with open("Saccharomyces_cerevisiae.R64-1-1.92.chromosome.II.gff3", "r") as fh:
    for line in fh:
        # Skip comments and headers
        if line.startswith("#"):
            continue

        fields = line.strip().split("\t")
        if len(fields) < 9:
            continue  # malformed line, skip

        ftype = fields[2]

        # Only work with "gene" features
        if ftype == "gene":
            chrom = fields[0]
            start = int(fields[3])
            end   = int(fields[4])
            attr  = fields[8]  # attributes column

            # Parse gene ID from the attributes field (look for "ID=")
            gene_id = None
            for item in attr.split(";"):
                if item.startswith("ID="):
                    gene_id = item.split("=", 1)[1]
                    break

            if gene_id is None:
                # If no ID= found, skip this entry
                continue

            gene_coords[gene_id] = (chrom, start, end)

print("Total gene entries found:", len(gene_coords))

# 2. Filter genes by length (> 2000 bp)
threshold = 2000

long_genes = {}
for gid, (chrom, start, end) in gene_coords.items():
    length = end - start + 1
    if length > threshold:
        long_genes[gid] = (chrom, start, end)

print("Number of genes longer than", threshold, "bp:", len(long_genes))

for i, k in long_genes.items():
    print(i,k)
````

| `d.copy()` | Copy dictionary (shallow) | `d2 = d.copy()` | Make safe working copy |

## Looping patterns for dictionaries

Looping through dictionaries is fundamental when processing real biological datasets. Below is a reference table of common looping patterns and when to use them.

| Loop Pattern | What It Does | Example Syntax | Bioinformatics Use |
|--------------|--------------|----------------|--------------------|
| Loop keys | Access each key | `for k in d:` | Iterate through sequence IDs |
| Loop values | Access each value | `for v in d.values():` | Scan read depths or counts |
| Loop key–value pairs | Cleanest pattern | `for k, v in d.items():` | Nucleotide counts, metadata tables |
| Loop sorted keys | Enforce order | `for k in sorted(d):` | Sorted sample names or loci |
| Nested loops | Work with hierarchical data | `for s, g in nested.items():` | Sample → gene → expression |
| Build dict in loop | Construct dictionary on the fly | `d[k]=v` inside loop | Build base counts, k-mer counts |
| Counting pattern (`get`) | Idiomatic Python counting | `d[x] = d.get(x,0)+1` | Count bases, alleles, codons |
| Grouping pattern (`setdefault`) | Build groups efficiently | `groups.setdefault(pop, []).append(id)` | Cluster samples by population |
